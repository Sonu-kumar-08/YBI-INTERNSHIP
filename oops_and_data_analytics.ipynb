```python
#encapsulation
#private cannot be accessed outside of class
#why do we need Encapsulation?
'''Protects data from unauthorized acess and accidental modification
  contols data updates using getter/setter method with validation
  enhances modulation by hiding internal implementation details
  Simplifies maintainance through centralized data handling logic
  '''
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary

    def get_salary(self):
        return self.__salary

emp = Employee("Fedrick", 5000)

print(emp.name)
print(emp.get_salary())
```

```python
class BankAccount:
    def __init__(self,balance):
      self._balance = balance

    def deposit(self,amount):
      self._balance += amount

    def withdraw(self,amount):
        if amount > 0 and amount <= self._balance:
            self._balance -= amount
        else:
          print("invalid withdrawl")
    def display_balance(self):
        print(f"balance: {self._balance}")
emp= BankAccount(50000)
emp.deposit(10000)
emp.withdraw(20000)
emp.display_balance()
```

```python
class Student:
  def __init__(self, Name ,Marks):
    self.Name = Name
    self.Marks = Marks

  def display(self):
    print(f"Name: {self.Name}")
    print(f"Marks: {self.Marks}")

  def Updatename(self, upName):
    self.Name=upName
    print(f"Updated Name: {self.Name}")

s= Student("Sonu",90)
s.Updatename("priyam")
s.display()
```

```python
class Student:
    def __init__(self, Name, Marks, regno):
        self.Name = Name
        self.Marks = Marks
        self._regno = regno

    def display(self):
        print(f"Name: {self.Name}")
        print(f"Marks: {self.Marks}")
        print(f"Regno: {self._regno}")

    def Updatemark(self, Upmark):
        self.PreviousMarks = self.Marks
        self.Marks = Upmark

        print(f"Previous marks: {self.PreviousMarks}")
        print(f"Updated marks: {self.Marks}")


s = Student("Sonu", 89, 101)

s.Updatemark(90)
s.display()
```

```python
#.polymorphism
#.Write reusable code
#.Reduce code complexity
#.make program flexible
#.easy add new classes
#.improve maintainability
#.use a common interface for different object
class Bird:
  def move(self):
    print("Bird moves by flying")

class Fish:
  def move(self):
    print("fish moves by swim")

class Car:
  def move(self):
    print("car moves by start")

object = [Bird(),Fish(),Car()]
for obj in object:
  obj.move()
```

```python
#method overrding
class Animal:
  def sound(self):
    print("Animal makes a sound")

class Dog(Animal):
  def sound(self):
    print("Dog barks")

class Cat(Animal):
  def sound(self):
    print("Cat meows")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

```python
#runtime Polymorphism using inheritance
class Vehical:

    def start(self):
        print("Vehicle starts")


class Car(Vehical):

    def start(self):
        print("Car starts with start button")


class Bike(Vehical):

    def start(self):
        print("Bike starts with start button")


vehi = Car()
vehi.start()

vehi = Bike()
vehi.start()
```
