```python
import numpy as np
```

```python
x = np.array([1,2,3,4])
y = x*2
plt.plot(x,y)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Simple line plot with lables")
plt.show()
```

```python
x = np.array([1,2,3,4,5])
y = [3,6,9,12,15]
plt.plot(x,y, marker = "o" , linestyle = '-' , label = 'Data Points')
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Simple line plot with lables")
plt.legend()
plt.show()
```

```python
x = np.array([1,2,3,4,5])
y = [3,6,9,12,15,]
plt.plot(x,y, marker = "s" , linestyle = '--' , label = 'Data Points')
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Simple line plot with lables")
plt.legend()
plt.show()
```

```python
x = np.array([1,2,3,4,5])
y = [3,6,9,12,15,]
plt.plot(x,y, marker = "s" , color = "purple", linestyle = '--' , label = 'Data points')
plt.grid()
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Simple line plot with lables")
plt.legend()
plt.show()
```

```python
 x = np.array([1,2,3,4,5])
 y = x*2

 x1=[2,4,6,8]
 y1=[3,5,7,9]

 plt.plot(x,y, linestyle = '--' , marker = 's',label = 's type marker')
 plt.plot(x1,y1, linestyle = '-.', marker = 'o', label = 'o type marker')
 plt.xlabel("X-axis")
 plt.ylabel("Y-axis")
 plt.title("Simple line plot with lables")
 plt.legend()
 plt.show()
```

```python
x = [1,2,3,4,5]
y = [3,6,9,12,15]


plt.figure(figsize=(8,6))
plt.plot(x,y, marker= 'o', linestyle='-')

for xi, yi in zip(x,y):
    plt.annotate(f'({xi},{yi})',
                  (xi,yi),
                  textcoords='offset points',
                  xytext=(0,10),
                 ha="center")

plt.title('Line Chart with Annotations')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid()
plt.show()
```

```python
 x = np.array([1,2,3,4])
 y = x*2
 y1=[3,5,7,9]

plt.plot(x, y, label = 'y= 2x')
plt.plot(x, y1, '-.', label = 'y1')
plt.fill_between(x,y,y1, color = 'red', alpha = 0.1)
plt.legend()

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Simple line plot with lables")
plt.savefig('line_plot.png')
plt.show()
```
